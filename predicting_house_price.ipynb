```python
#doing linear regression
x_train=np.array([1.0,2.0])  #it is the input variable(in 1000sq feets)
y_train=np.array([100,300])   # it is the output variable in 1000$s'
```

```python
#no of training m
m=len(x_train)
print("No. of training examples: "+str(m))
```

```python
i=0 # changing x_i and y_i inot x^i,y^i
for i in range(m):
    x_i=x_train[i]
    y_i=y_train[i]
    print("x_i = "+str(x_i)+" y_i = "+str(y_i))
print(x_train.shape)
print(np.zeros(x_train.shape))
```

```python
#plot the points
plt.xlabel("Size in 1000sq feets")
plt.ylabel("Prices in 1000$s'")
plt.title("Housing Prices")
for i in range(m):
    plt.scatter(x_train[i],y_train[i],marker='+',c='r')
plt.show()
```

```python
w,b=100,100
print('w: '+str(w)+' b: '+str(b))
def cal_f_w(w,b,x):
    m=x.shape
    n=len(x)
    f_wb=np.zeros(m) # gives a list of zeros with 'm' length
    for i in range(n):
        f_wb[i]=w*x[i]+b
    return f_wb    
```

```python
tmp_f_wb=cal_f_w(w,b,x_train)
plt.plot(x_train,tmp_f_wb,c='b',label='Our Prediction') #plot(x_coordinates,y_coordinates)
plt.xlabel("Size of house in 1000sq. ft")
plt.ylabel("Price of house in 1000$s'")
plt.scatter(x_train,y_train,c='r',marker='+',label='Actual Value')
plt.legend() #use to specify the area where elements need to be shown
new_x=np.array([1.5])
tmp=cal_f_w(w,b,new_x)
plt.scatter(new_x,tmp,marker='+',c='g')
plt.title("Housing Prices")
plt.show()
```
