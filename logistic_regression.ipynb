```python
import numpy as np
from numpy import random
import csv
```

```python
x=[]
y=[]
with open('C:/Users/DHRUVRAJ/Downloads/framingham.csv','r') as csv_file:
    csv_reader=csv.reader(csv_file)
    for line in csv_reader:
        x.append(line[0:15])
        y.append(line[15])
x.remove(x[0])
y.remove(y[0])
x_array=np.array(x,float)
y_array=np.array(y,float)
x1=x_array.reshape(-1,15)
y1=y_array.reshape(-1,1)
```

```python
def sigmoid(x):
    return 1/(1+np.exp(-x)) 
```

```python
#creating linear regression moodel:

#hyper parameters---->parameters decided by ourselves
#model parameters---->parameters decided by the model
class Logistic_Regression:
    #initating the parameters
    def __init__(self,learning_rate=0.001,no_of_iterations=1000): 
        self.learning_rate=learning_rate
        self.no_of_iterations=no_of_iterations
        
    def fit(self,X,Y):# X-->years of experience  Y-->amount of salary
        #no of training examples(m)  and no. of features(n)
        self.m,self.n=X.shape #no of rows==self.m and columns=self.n
        #initating the weight and bias
        self.w=np.zeros((1,self.n)) #each wewight for each feature and since no_of features is equal ot self.n
        self.b=0
        self.X=X
        self.Y=Y

    #impplementing Gradient descent
        for i in range(self.no_of_iterations):
            self.update_weights()

    
    def update_weights(self):
        Y_prediction=self.predict(self.X)
        #CALCULATE GRADIENT
        dw=-(((self.X.T).dot(self.Y-Y_prediction)))/self.m
        db=-(np.sum(self.Y-Y_prediction))/self.m
        #update weights
        self.w=self.w-self.learning_rate*dw.T
        self.b=self.b-self.learning_rate*db
        

    def predict(self,X):
        m=X.shape[0]
        lr_pred=X.dot(self.w.T)+self.b #dot product of vector X and w
        y_pred=sigmoid(lr_pred)
        l=[]
        for y in y_pred:
            if y>=0.5:
                l.append('1')
            else:
                l.append('0')
        arr=np.array(l,int)
        l_new=arr.reshape((m,1))
        return l_new
       
        
```

```python
model=Logistic_Regression()

model.fit(x1,y1)

a=np.array([[0,41,3,0,0,1,0,1,0,332,124,88,31.31,65,84],[1,39,4,0,0,0,0,0,0,195,106,70,26.97,80,77],[0,61,3,1,30,0,0,1,0,225,150,95,28.58,65,103]])
model.predict(a)
```
